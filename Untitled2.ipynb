Відокремлення кренів аналітично
$f(x) = 2x^2+4x^3-x^2-3x-1=0$

```python
import numpy as np

def f(x):
    return 2*x**4 + 4*x**3 - x**2 - 3*x - 1

eps = 0.0001

def find_segments():
    search_range = np.arange(-10, 11, 1)
    segments = []
    prev_x = search_range[0]
    prev_f = f(prev_x)
    for x in search_range[1:]:
        cur_f = f(x)
        if prev_f * cur_f < 0:
            segments.append((prev_x, x))
        prev_x, prev_f = x, cur_f
    return segments

def dichotomy(a, b, eps):
    while abs(b - a) > eps:
        mid = (a + b) / 2
        if f(a) * f(mid) < 0:
            b = mid
        else:
            a = mid
    return (a + b) / 2

def chords(a, b, eps):
    fa, fb = f(a), f(b)
    if fa * fb > 0:
        raise ValueError("На відрізку [{}, {}] немає зміни знаку".format(a, b))
    x_old = a
    while True:
        x = (a*fb - b*fa) / (fb - fa)
        fx = f(x)
        if abs(x - x_old) < eps or abs(fx) < eps:
            return x
        if fa * fx < 0:
            b, fb = x, fx
        else:
            a, fa = x, fx
        x_old = x

segments = find_segments()
print("Відокремлені відрізки з коренями:", segments)

for a, b in segments:
    root_bis = dichotomy(a, b, eps)
    root_chords = chords(a, b, eps)
    print(f"Інтервал [{a}, {b}]")
    print(f"  Метод половинного ділення: x ≈ {round(root_bis, 5)}")
    print(f"  Метод хорд: x ≈ {round(root_chords, 5)}")
```
